**Question:** A first case appears in a population where nobody is immune. For which contact and recovery rates does it start an outbreak at all, and what share of the population has to be immune beforehand for the outbreak to fail?

## Method

The basic reproduction ratio $R_0$ is the number of people that one infected person infects, on average, in a population where everybody is susceptible (Diekmann, Heesterbeek and Metz, 1990). In the SIR model an infected person meets $\beta$ people per day and stays infectious for $1 / \gamma$ days, so $R_0 = \beta / \gamma$. At the start $S \approx N$, the equation for the infected becomes $dI/dt \approx (\beta - \gamma) I$, and $I$ grows exponentially at rate $\gamma (R_0 - 1)$ when $R_0 > 1$ and decays when $R_0 < 1$. For the compartmental models and hypotheses considered by the cited work, $R_0 < 1$ gives local disease-free stability, not an unconditional global-eradication result (van den Driessche and Watmough, 2002).

In the negligible-seed approximation, if a share $v$ is immune before the outbreak, an infected person meets susceptible people only $(1 - v)$ of the time and the effective ratio is $R_0 (1 - v)$. It drops below 1 when $v > 1 - 1/R_0$, the herd threshold. The engine has no immune compartment, but immunising a share $v$ is the same as running it on a population of $(1 - v) N$ people with the contact rate $\beta (1 - v)$, because the infection pressure $\beta / N$ then stays unchanged. The notebook uses that equivalence.

The first cell prints $R_0$ for the five cases on file and, where it exceeds 1, the herd threshold. It runs no simulation; it is the arithmetic the rest of the notebook will test against the engine.

````python
# the five cases with R0 = beta / gamma and, where R0 is above 1, the herd threshold 1 - 1 / R0
for case_id in CASES:
    p = params_of(case_id)
    r0 = p.beta / p.gamma
    threshold = f"{1 - 1 / r0:.3f}" if r0 > 1 else "none (R0 <= 1)"
    print(f"{case_id:18s} R0 = {r0:5.2f}   herd threshold: {threshold}")
````

Only the sub-critical case, at 0.72, sits below 1. The control has $R_0 = 2.00$ but nobody infected to start with, so there is nothing to grow, and the other three have $R_0$ of 1.20, 2.75 and 6.00. The next cell runs the engine on every case that has infected people and divides the peak by the number infected on day 0. If $R_0$ is the threshold, the sub-critical case should not grow at all and the others should grow by a large factor.

````python
# run the cases that have infected people and set each peak against the people infected at the start
for case_id in CASES:
    p = params_of(case_id)
    if p.I0 == 0:
        continue
    r = simulate(p)
    print(f"{case_id:18s} R0 = {p.beta / p.gamma:5.2f}  I0 = {p.I0:4.0f}", end="  ")
    print(f"peak {r.peak_I:7.0f} on day {r.t_peak:6.2f}  x{r.peak_I / p.I0:6.1f}")
````

The sub-critical case multiplies its 50 initial infections by exactly 1.0: its peak is day 0. The slow spread, with an $R_0$ only 20 percent above 1, multiplies them by 30, and the epidemic and the fast burn by about 550. So the threshold shows, but three cases are too few to see its shape. The next cell sweeps $R_0$ from 0.5 to 4 in steps of 0.1, with the recovery rate of the epidemic case fixed and the contact rate changed, and runs the engine at each value. For each run it fits a straight line to the logarithm of the infected people over the first week, whose slope is the observed growth rate, and records the peak as a share of the population. It prints four values next to the theory: the rate $\gamma (R_0 - 1)$ and the analytic peak of the baseline.

````python
# sweep R0 from 0.5 to 4 with the recovery rate of the epidemic case, and run the engine at every value
base = params_of("EX02_epidemic")
r0_grid = np.linspace(0.5, 4.0, 36)
growth = []
peak_share = []
for r0 in r0_grid:
    r = simulate(replace(base, beta=r0 * base.gamma))
    growth.append(np.polyfit(r.t[:29], np.log(r.I[:29]), 1)[0])
    peak_share.append(r.peak_I / base.N)
baseline = [peak_frac(r0) for r0 in r0_grid]
for r0 in (0.5, 1.0, 2.0, 4.0):
    k = int(np.argmin(np.abs(r0_grid - r0)))
    print(f"R0 = {r0:3.1f}  growth {growth[k]:+.3f}  gamma (R0 - 1) = {base.gamma * (r0 - 1):+.3f}", end="  ")
    print(f"peak share {peak_share[k]:.4f}  analytic {baseline[k]:.4f}")
````

The next cell draws the growth rate and the peak share against $R_0$ in two panels, with the threshold $R_0 = 1$ marked.

````python
# draw the first-week growth rate and the peak share against R0, with the threshold R0 = 1 marked
fig, (left, right) = plt.subplots(1, 2, figsize=(9, 3.8))
left.plot(r0_grid, growth, "o", label="engine")
left.plot(r0_grid, base.gamma * (r0_grid - 1), "k:", label="gamma (R0 - 1)")
right.plot(r0_grid, peak_share, "o", label="engine")
right.plot(r0_grid, baseline, "k:", label="analytic peak")
ylabels = ("first-week growth rate (1/day)", "peak infected share of the population")
for ax, ylabel in zip((left, right), ylabels):
    ax.axvline(1.0, color="grey", linewidth=0.8)
    ax.set_xlabel("R0 = beta / gamma")
    ax.set_ylabel(ylabel)
    ax.legend()
fig.suptitle("Take-off against R0: the epidemic case with beta varied")
fig.tight_layout()
plt.show()
````

The left panel is the threshold itself: the first-week growth rate crosses close to zero at $R_0 = 1$ and follows the line $\gamma (R_0 - 1)$, a little below it at large $R_0$ because by the end of the week the susceptible people have already started to run out. The right panel shows how little a high rate matters just above the threshold: the peak reaches about 1.52 percent of the population at $R_0 = 1.2$ and then climbs steeply, tracking the analytic peak of the baseline. The figure does not show time. An outbreak with $R_0$ just above 1 does take off, but how late is what the next cell shows.

````python
# look closely around the threshold: the engine at R0 = 0.9, 1.0, 1.1, 1.2 and 1.5
for r0 in (0.9, 1.0, 1.1, 1.2, 1.5):
    r = simulate(replace(base, beta=r0 * base.gamma))
    print(f"R0 = {r0:3.1f}  peak {r.peak_I:6.0f}  on day {r.t_peak:6.2f}  removed fraction {r.attack_rate:.4f}")
    print(f"peak as percent of N={100*r.peak_I/base.N:.2f}%")
````

At $R_0 = 0.9$ and at $R_0 = 1.0$ the peak is the 50 initial infections on day 0. At $R_0 = 1.1$ the infected are still growing when the 160 days end: the peak printed, on day 160.00, is the last day and not a true peak. At 1.2 the peak comes at day 120.25 and at 1.5 at day 64.50. Close above the threshold the outbreak exists but is slow, so a horizon that is too short mistakes it for no outbreak. The last question of the notebook is the herd threshold: the cell runs the epidemic case with a share $v$ of the population immune, from 0 to 0.8 in steps of 0.05, and prints the threshold $1 - \gamma/\beta$ with the infected share for three values of $v$.

````python
# Keep removed and ever-infected fractions separate under initial immunity.
shares = np.linspace(0.0, 0.8, 17)
final_share = []
ever_share = []
for v in shares:
    scaled = replace(base, beta=base.beta * (1 - v), N=base.N * (1 - v))
    immune_run = simulate(scaled)
    final_share.append(immune_run.R[-1] / base.N)
    ever_share.append((immune_run.I[-1] + immune_run.R[-1]) / base.N)
herd = 1 - base.gamma / base.beta
finite_herd = herd - base.I0 / base.N
print(f"small-seed threshold={herd:.6f}, finite-seed initial-growth threshold={finite_herd:.6f}")
for v in (0.3, 0.6, 0.7):
    k = int(np.argmin(np.abs(shares - v)))
    print(f"v={v:.1f}, removed R(T)/N={final_share[k]:.6f}, ever infected={ever_share[k]:.6f}")
````

The next cell draws the infected share against the immune share, with the herd threshold marked.

````python
# draw the fraction of the original population by T against the immune share, with the herd threshold marked
fig, ax = plt.subplots(figsize=(7, 3.8))
ax.plot(shares, final_share, "o-", label="removed R(T)/N")
ax.plot(shares, ever_share, "s--", label="ever infected [I(T)+R(T)]/N")
ax.axvline(herd, color="grey", linestyle=":", label=f"small-seed threshold {herd:.3f}")
ax.set_xlabel("immune share v before the outbreak")
ax.set_ylabel("fraction of the original population by T")
ax.set_title("EX02_epidemic: immunised at the start")
ax.legend()
fig.tight_layout()
plt.show()
````

The two curves distinguish removed R(T)/N from everyone infected by T, [I(T)+R(T)]/N. While the outbreak is unfinished, their vertical separation represents people still infected rather than conservation error. For v=0.3/0.6/0.7 the values 0.5446/0.0496/0.0028 belong to the removed curve, divided by the original population. The vertical line uses the vanishing-seed threshold; the printed calculation also gives the finite-initial-state correction. Perfect uniform immunity remains an authored assumption, not evidence of the effectiveness of a real intervention.

### The initially susceptible fraction sharpens the threshold

The growth criterion follows directly from the infected equation, without fitting a trajectory:

$$\frac{1}{I}\frac{dI}{dt}=\gamma\left(R_0\frac{S}{N}-1\right),\qquad R_{\rm eff}(0)=R_0(1-v-i_0).$$

Here $v$ is initially immune share of the original population and $i_0=I_0/N$ its infected share; the shares must leave a nonnegative susceptible population. Setting $v=0$ and treating $i_0$ as negligible recovers the familiar threshold one. With a nonzero seed, the exact initial-growth condition uses the actual $S(0)/N$, so the plotted first-week rate only approximately crosses zero at one. First-week regression also averages susceptible depletion and Euler error, whereas the equation gives an instantaneous derivative.

For the immunity experiment, scaling population and contact rate preserves infection pressure on the remaining nonimmune people. The output is then divided by the original population, which avoids confusing a fraction of the remaining group with a fraction of the whole town. Perfect initial immunity is a constructed SIR scenario, not vaccine effectiveness estimated from observations. The zero-infection control is another distinct limit: a positive reproduction ratio cannot create an infection in an exactly unseeded deterministic equation.

## Results

**Answer:** With a positive seed, instantaneous growth requires R0 S(0)/N>1. The approximation R0≈1 assumes an initially almost susceptible population; first-week regression also includes susceptible depletion and Euler error. At R0=1.2 the peak is about 1520 people, roughly 1.52 percent rather than only 1 percent. For the epidemic case the vanishing-seed immunity threshold is 0.636364, while the initial-growth threshold with 50 infected people is 0.635864.

The immunity experiment's0.5446,0.0496 and 0.0028 for v=0.3/0.6/0.7 refer to removed fraction R(T)/N. The added column and curve show [I(T)+R(T)]/N, which is larger whenever infected people remain. Both fractions use the original population, not the reduced nonimmune group. Uniform perfect immunity and homogeneous contact are authored assumptions, not operational findings or a global guarantee for arbitrary initial states.

## Summary

- Instantaneous growth uses the actual initial susceptible fraction and a nonzero seed; R0≈1 is the nearly susceptible limit.
- First-week growth estimates approximate rather than exactly equal the initial derivative.
- At R0=1.2 the peak is about 1.52 percent; near the threshold a short horizon can conceal its timing.
- Removed and ever-infected fractions at a cutoff are different quantities.
